# Validação do modelo YOLO salvo (versão independente)
Carrega o `best.pt` treinado e mede top-1 em **valid** e **test**, sem depender de variáveis de outro notebook nem do Google Drive.
Coloque este arquivo na pasta `Treinamento_YOLO`.

```python
from pathlib import Path
from ultralytics import YOLO

BASE   = Path.cwd()
PESO   = BASE / "runs" / "classify" / "runs" / "yolo11s_lung_v2" / "weights" / "best.pt"
DADOS  = BASE / "Tumor-Cerebral-1"   # pasta com train/valid/test
ALVO   = 512

assert PESO.exists(),  f"Peso não encontrado: {PESO}"
assert DADOS.exists(), f"Dados não encontrados: {DADOS}"

modelo = YOLO(str(PESO))
r_val  = modelo.val(data=str(DADOS), split="val",  imgsz=ALVO)
r_test = modelo.val(data=str(DADOS), split="test", imgsz=ALVO)
print("Validação top1:", r_val.top1, "| Teste top1:", r_test.top1)
```
